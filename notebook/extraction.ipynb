```python

print("LES INCOHERENCES : ")
print("Age :")
print(df[(df["Age"] < 0) | (df["Age"] > 100)])
print("Sleep Hours :")
print(df[(df["Sleep_Hours"] < 0) | (df["Sleep_Hours"] > 24)])

print("Study Hours :")
print(df[
    (df["Study_Hours_Per_Week"] < 0) |
    (df["Study_Hours_Per_Week"] > 44)
])

print("GPA :")
print(df[(df["GPA"] < 0) | (df["GPA"] > 4)])
for col in ["Social_Media_Hours", "Screen_Time_Hours"]:
    print(f"\n{col} :")
    print(df[(df[col] < 0) | (df[col] > 24)])

print("Attendance Rate :")
print(df[
    (df["Attendance_Rate"] < 0) |
    (df["Attendance_Rate"] > 100)
])

for col in ["Financial_Stress", "Family_Support"]:
    print(f"\n{col} :")
    print(df[(df[col] < 0) | (df[col] > 10)])

print("\nYear of Study :")
print(df[
    (df["Year_of_Study"] < 1) |
    (df["Year_of_Study"] > 5)
])
```

Les valeurs aberrantes :

```python
%matplotlib inline

import matplotlib.pyplot as plt
import seaborn as sns

numeric_columns = [
    "Age",  
    "Sleep_Hours",
    "Study_Hours_Per_Week",
    "GPA",
    "Social_Media_Hours",
    "Screen_Time_Hours",
    "Attendance_Rate",
    "Financial_Stress",
    "Family_Support",
    "Year_of_Study"
]  

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | 
                  (df[col] > upper_bound)]
    
    print(f"Outliers in {col}:")
    print("le nombre d'outliers :", len(outliers))
    print("Borne inferieure :", lower_bound)
    print("Borne superieure :", upper_bound)
```

Visualisation d'outliers

```python
%matplotlib inline

import matplotlib.pyplot as plt
import seaborn as sns

for col in numeric_columns:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```
