Les valeurs aberrantes :

```python
%matplotlib inline

import matplotlib.pyplot as plt
import seaborn as sns

numeric_columns = [
    "Age",  
    "Sleep_Hours",
    "Study_Hours_Per_Week",
    "GPA",
    "Social_Media_Hours",
    "Screen_Time_Hours",
    "Attendance_Rate",
    "Financial_Stress",
    "Family_Support",
    "Year_of_Study"
]  

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | 
                  (df[col] > upper_bound)]
    
    print(f"Outliers in {col}:")
    print("le nombre d'outliers :", len(outliers))
    print("Borne inferieure :", lower_bound)
    print("Borne superieure :", upper_bound)
```

```python

# 1. Encoder temporairement la variable cible

df['Mental_Health_Risk_Encoded'] = df['Mental_Health_Risk'].map({
    'Low': 0,
    'Medium': 1,
    'High': 2
})

# 2. Sélectionner les variables numériques et la cible

correlation_columns = [
    'Age',
    'Sleep_Hours',
    'Study_Hours_Per_Week',
    'GPA',
    'Social_Media_Hours',
    'Screen_Time_Hours',
    'Attendance_Rate',
    'Mental_Health_Risk_Encoded'
]

# 3. Calculer la matrice de corrélation

correlation_matrix = df[correlation_columns].corr()

# 4. Afficher la Heatmap

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap - Mental Health Risk')

plt.tight_layout()
plt.show()
```

### Identification des variables les plus corrélées au risque de santé mentale

L'analyse de la matrice de corrélation montre que la variable **Sleep_Hours** présente la corrélation la plus importante avec le risque de santé mentale (-0.54), suivie de **GPA** (-0.35) et de **Study_Hours_Per_Week** (-0.24).

Les variables **Social_Media_Hours** (+0.21) et **Screen_Time_Hours** (+0.17) présentent des corrélations positives faibles avec le risque.

En revanche, **Age** présente une corrélation presque nulle (0.00).

Ainsi, selon cette analyse exploratoire, les heures de sommeil et le GPA sont les variables qui présentent les associations linéaires les plus fortes avec le niveau de risque dans ce dataset.

*Remarque :
 ces corrélations indiquent des associations et ne prouvent pas une relation de causalité. L'encodage Low = 0, Medium = 1 et High = 2 permet une analyse exploratoire, mais les coefficients de Pearson restent approximatifs pour une cible catégorielle ordinale.*
